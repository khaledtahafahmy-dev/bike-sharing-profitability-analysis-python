```python
# =====================================================
# 4. Data Cleaning
# =====================================================
df_yr0 = df_yr0.rename(columns={
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "hr": "hour",
    "hum": "humidity",
    "weathersit": "weather_condition"
})

df_yr1 = df_yr1.rename(columns={
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "hr": "hour",
    "hum": "humidity",
    "weathersit": "weather_condition"
})
```

```python
display(df_yr0.columns)
display(df_yr1.columns)
```

```python
df_yr0["date"] = pd.to_datetime(
    df_yr0["date"],
    format="%d/%m/%Y"
)

df_yr1["date"] = pd.to_datetime(
    df_yr1["date"],
    format="%d/%m/%Y"
)
```

```python
display(df_yr0.dtypes)
display(df_yr1.dtypes)
```

```python
cost_table.rename(columns={"yr": "year"}, inplace=True)
```

```python
df_yr0 = df_yr0.merge(cost_table, on="year", how="left")
```

```python
df_yr1 = df_yr1.merge(cost_table, on="year", how="left")
```

```python
display(df_yr0.head(11))
display(df_yr1.head(11))
```

```python
# =====================================================
# 5. Data Transformation
# =====================================================
df_yr0["profit_per_ride"] = df_yr0["price"] - df_yr0["COGS"]
df_yr1["profit_per_ride"] = df_yr1["price"] - df_yr1["COGS"]
```

```python
df_yr0["revenue"] = df_yr0["riders"] * df_yr0["price"]
df_yr1["revenue"] = df_yr1["riders"] * df_yr1["price"]
```

```python
df_yr0["total_cost"] = df_yr0["riders"] * df_yr0["COGS"]
df_yr1["total_cost"] = df_yr1["riders"] * df_yr1["COGS"]
```

```python
df_yr0["total_profit"] = df_yr0["revenue"] - df_yr0["total_cost"]
df_yr1["total_profit"] = df_yr1["revenue"] - df_yr1["total_cost"]
```

```python
df_all = pd.concat([df_yr0, df_yr1], ignore_index=True)
```

```python
display(df_all.info())
display(df_all.head())
display(df_all.describe())
```

```python
season_map = {
    1: "Winter",
    2: "Spring",
    3: "Summer",
    4: "Fall"
}

df_all["season_name"] = df_all["season"].map(season_map)
```

```python
weather_map = {
    1: "Clear",
    2: "Mist",
    3: "Light Snow / Rain",
    4: "Heavy Rain / Snow"
}

df_all["weather_name"] = df_all["weather_condition"].map(weather_map)
```

```python
year_map = {
    0: 2021,
    1: 2022
}

df_all["year"] = df_all["year"].map(year_map)
```

```python
df_all["weekday_name"] = df_all["date"].dt.day_name()
```

```python
df_all["month_name"] = df_all["date"].dt.month_name()
```

```python
holiday_map = {
    0: "No",
    1: "Yes"
}

df_all["holiday"] = df_all["holiday"].map(holiday_map)
```

```python
workingday_map = {
    0: "Non-Working Day",
    1: "Working Day"
}

df_all["workingday"] = df_all["workingday"].map(workingday_map)
```

```python
display(df_all.head(11))
display(df_all.tail(11))
```

```python
# =====================================================
# 6. Exploratory Data Analysis (EDA)
# =====================================================
profit_trend = df_all.pivot_table(
    values="total_profit",
    index="month_name",
    columns="year",
    aggfunc="sum"
)

display(profit_trend)
```

```python
weather_profit = df_all.pivot_table(
    values="total_profit",
    index="weather_name",
    aggfunc="sum"
)

display(weather_profit)
```

```python
hour_profit = df_all.pivot_table(
    values="total_profit",
    index="hour",
    aggfunc="sum"
)

display(hour_profit)
```

```python
rider_profit = df_all.pivot_table(
    values="total_profit",
    index="rider_type",
    aggfunc="sum"
)

display(rider_profit)
```

```python
# =====================================================
# 7. Visualization
# =====================================================
plt.figure(figsize=(12,6))

sns.lineplot(
    data=profit_trend,
    markers=True,
    dashes=False,
    linewidth=3
)

plt.title("Monthly Profit Trend (2021 vs 2022)", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
weather_profit =df_all.pivot_table(
    values="total_profit",
    index="weather_name",
    aggfunc="sum"
).sort_values("total_profit", ascending=False)

plt.figure(figsize=(8,5))

sns.barplot(
    data=weather_profit,
    x=weather_profit.index,
    y="total_profit"
)

plt.title("Profit by Weather Condition", fontsize=16)
plt.xlabel("Weather")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

```python
hour_profit = df_all.pivot_table(
    values="total_profit",
    index="hour",
    aggfunc="sum"
)

plt.figure(figsize=(12,5))

sns.lineplot(
    data=hour_profit,
    x=hour_profit.index,
    y="total_profit",
    marker="o",
    linewidth=3
)

plt.title("Profit by Hour of Day", fontsize=16)
plt.xlabel("Hour")
plt.ylabel("Total Profit ($)")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
rider_profit = df_all.pivot_table(
    values="total_profit",
    index="rider_type",
    aggfunc="sum"
)

plt.figure(figsize=(6,5))

sns.barplot(
    data=rider_profit,
    x=rider_profit.index,
    y="total_profit"
)

plt.title("Profit by Rider Type", fontsize=16)
plt.xlabel("Rider Type")
plt.ylabel("Total Profit ($)")

plt.tight_layout()
plt.show()
```
